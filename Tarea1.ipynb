# Cambie una lista a un una tupla, a un diccionario y a un conjunto.

```python
lista = ["Naranja", "Mango", "Coco", "Fresa", "Limón", "Manzana"]
```

```python
tupla = tuple(lista)
tupla
```

```python
diccionario = dict(enumerate(lista))
diccionario
```

```python
conjunto = set(lista)
conjunto
```
