filter es una función integrada de python. Filtra valores de una lista dada una condición boleana.

```python
personas = [("Juan", 20), ("Ana", 18), ("Luis", 25)]
def clave(diccionario):
  return sorted([i for i in diccionario])
```

```python
clave(personas)
```

```python
ordenado = sorted(personas, key= lambda persona: persona)
ordenado
```

```python
datos = [("a", 3), ("b", 1), ("c", 2)]

ordenado = sorted(datos, key=lambda x: x[0])
print(ordenado)
```

```python
datos = [("a", 3), ("b", 1), ("c", 2)]

ordenado = sorted(datos, key=lambda x: x[1])
print(ordenado)
```

```python
datos = [("a", 3, "x"), ("b", 1, "z"), ("c", 2, "y")]
ordenado = sorted(datos, key= lambda x: x[2])
print(ordenado)
```

```python
def longitud(palabra):
  return len(palabra)
```
