```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
import io
f_capitulos1 = io.BytesIO(uploaded['f_capitulos.csv'])
f_capitulos1 = pd.read_csv(f_capitulos1, sep = ",", encoding = "latin-1")
f_capitulos1.head(4)
```

```python
tabla_alcohol = pd.DataFrame({
    'edad': f_capitulos1['F_04'],
    'borrachera': f_capitulos1['F_09'],
    'cerveza': f_capitulos1['F_10_A'],
    'recordar': f_capitulos1['F_19'],
    'herido': f_capitulos1['F_20_CUAL'],
    'preocupacion': f_capitulos1['F_21_CUAL'],
    'familia': f_capitulos1["F_01_CUAL"],
    'amigos': f_capitulos1['F_02_CUAL'],
})
tabla_alcohol.head(5)
```

```python
tabla_alcohol['cerveza'] = tabla_alcohol['cerveza'].fillna(0)
tabla_alcohol.head(5)
```

```python
tabla_alcohol['borrachera'] = tabla_alcohol['borrachera'].fillna(0)
tabla_alcohol.head(5)
```

```python
tabla_alcohol['recordar'] = tabla_alcohol['recordar'].fillna(0)
tabla_alcohol.head(5)
```

```python
tabla_alcohol['herido'] = tabla_alcohol['herido'].fillna(0)
tabla_alcohol.head(5)
```

```python
tabla_alcohol['preocupacion'] = tabla_alcohol['preocupacion'].fillna(0)
tabla_alcohol.head(5)
```

```python
tabla_alcohol['familia'] = tabla_alcohol['familia'].fillna(0)
tabla_alcohol.head(5)
```

```python
tabla_alcohol['amigos'] = tabla_alcohol['amigos'].fillna(0)
tabla_alcohol.head(5)
```

```python
tabla_alcohol1 = tabla_alcohol.dropna()
tabla_alcohol1.head(5)
```

```python
tabla_alcohol1['cerveza'] = tabla_alcohol1['cerveza'].astype(str)
tabla_alcohol1['recordar'] = tabla_alcohol1['recordar'].astype(str)
tabla_alcohol1['herido'] = tabla_alcohol1['herido'].astype(str)
tabla_alcohol1['familia'] = tabla_alcohol1['familia'].astype(str)
tabla_alcohol1['amigos'] = tabla_alcohol1['amigos'].astype(str)
tabla_alcohol1['preocupacion'] = tabla_alcohol1['preocupacion'].astype(str)
```
